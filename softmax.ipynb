```python
import torch
```

```python
import torch.nn as nn
```

```python
import numpy as np
```

```python
def softmax(x):
    return np.exp(x)/np.sum(np.exp(x),axis=0)
```

```python
x=np.array([2.0,1.0,0.1])
```

```python
outputs=softmax(x)
```

```python
print('softmax =',outputs)
```

```python
x=torch.tensor([2.0,1.0,0.1])
```

```python
outputs=torch.softmax(x,dim=0)
```

```python
print(outputs)
```

```python
loss=nn.CrossEntropyLoss()
```

```python
y=torch.tensor([0])
y_pred_good=torch.tensor([[2.0,1.0,.1]])
```
