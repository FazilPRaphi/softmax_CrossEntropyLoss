```python
from torch.utils.data import Dataset,DataLoader
```

```python
import numpy as np
```

```python
import math
```

```python
class Wine(Dataset):
    def __init__(self):
        xy=np.loadtxt('wine.csv',delimiter=",",dtype=np.float32,skiprows=1)
        self.x=torch.from_numpy(xy[:,1:])
        self.y=torch.from_numpy(xy[:,[0]])
        self.n_samples=xy.shape[0]
    def __getitem__(self,index):
        return self.x[index],self.y[index]
    def __len__(self):
        return self.n_samples
data=Wine()
dataloader=DataLoader(dataset=data,batch_size=4,shuffle=True,num_workers=0)
first_data=data[0]
features,labels=first_data
print(features,labels)
        
```

```python
num_epochs=2
```

```python
total_samples =len(data)
n_iteration=math.ceil(total_samples/4)
print(total_samples,n_iteration)
```

```python
for epoch in range(num_epochs):
    for i, (inputs,labels) in enumerate(dataloader):
        if (i+1)%5==0:
            print(f'epoch {epoch+1}/{num_epochs},step {i+1}/{n_iteration},inputs={inputs.shape}')
```
